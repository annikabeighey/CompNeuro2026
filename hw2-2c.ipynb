```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def B(V,Mg):
    return 1 / (1 + (Mg / 3.57) * np.exp(-V / 16.13))
```

```python
xvals = np.linspace(-90,0,600)

#Mg = 0
yvals0 = B(xvals,0)*xvals

#Mg = 0.1
yvals01 = B(xvals,0.1)*xvals

#Mg = 1
yvals1 = B(xvals,1)*xvals
```

```python
plt.plot(xvals,yvals0,label = 'Mg = 0 mM',linestyle='-')
plt.plot(xvals,yvals01,label = 'Mg = 0.1 mM',color = 'green')
plt.plot(xvals,yvals1,label = 'Mg = 1 mM', color = 'purple')
plt.xlabel('Voltage (mV)')
plt.ylabel('I/g (mV)')
plt.title('Current vs. Voltage for NMDA receptor')
plt.legend()
plt.tight_layout()
plt.grid(True, alpha=0.3)
plt.show()
```

For $Mg=0$, it is just a straight line because voltage and current are directly proportional. The slope is constant and positive. As voltage increases, so does current.

For $Mg=0.1$, when voltage is below about $-45$ mV the slope is negative, so an increase in voltage decreases current. The inhibition and leakiness start working when the voltage is increased, resulting in a decrease in current. Above $-45$ mV, the slope is positive because the exictation conductance has overcome conductance of the inhibition and leaky.

For $Mg=1$, it has a similar pattern to $Mg=0.1$ except the crucial voltage value is about $-25$ mV. Additionally, the slopes are not as steep. This is because magenesium blocks the NMDA receptor from working unless there is a positive change in voltage. When there was an increase in voltage the receptor opened to let current flow, which is seen in the $Mg=0.1$ example as well. However with a higher concentration of magnesium in the synapse, it is more likely that magnesium doesn't stop blocking the receptor fully or as much, even with a change in voltage. This is why the change in current is less steep for the $Mg=1$. The critical voltage is less negative because more of a positive charge will be needed to displace the larger amount of magnesium so that the receptor can let current into the cell.
