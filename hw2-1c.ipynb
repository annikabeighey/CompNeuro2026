**Question 1, part c**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def c(a):
    num = 1-2*a
    denom = np.sqrt(a*(1-a))
    return num/denom
```

```python
xvals = np.linspace(0,1,100)
yvals = c(xvals)
```

```python
plt.plot(xvals,yvals,label = 'active speed')
plt.axhline(2,label = 'passive apparent speed',color = 'green')
plt.axvline(0.5-np.sqrt(2)/4,label = 'crossing',color = 'red')
plt.axvline(0.5,label= 'a=1/2',color = 'lightpink')
plt.xlabel('a from 0 to 1 (excitation threshold)')
plt.ylabel('c(a) (speed of active front)')
plt.title('c as a function from 0 to 1 at D=1')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
